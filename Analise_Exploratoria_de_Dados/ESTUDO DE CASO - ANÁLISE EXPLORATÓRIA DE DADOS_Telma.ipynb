### **QUESTÃO 2**

#### Dentro de cada seção, quais categorias de produtos mais contribuíram com o faturamento bruto (produto entre a quantidade de unidades vendidas e o preço unitário)?

```python
# Insira seus códigos em Python aqui (divida em várias células, se desejar)
dados['FATURAMENTO_BRUTO'] = dados['QTD_UNIDADES_VENDIDAS_MES'] * dados['PRECO_UNITARIO']
dados.head(10)
```

```python

dados_1 = dados.groupby(["SECAO_PRODUTO", "CATEGORIA_PRODUTO"], as_index=False) \
                .agg(FATURAMENTO_BRUTO_TOTAL=("FATURAMENTO_BRUTO", "sum")) \
                .sort_values(by=["SECAO_PRODUTO", "FATURAMENTO_BRUTO_TOTAL"], ascending=[True, False])
```

```python
dados_1.head(10)
```

```python
dados_1["rank"] = dados_1.groupby("SECAO_PRODUTO")["FATURAMENTO_BRUTO_TOTAL"].rank(method="first", ascending=False)

dados_1.head(100)
```

```python
dados_2 = dados_1.sort_values(by=["SECAO_PRODUTO", "FATURAMENTO_BRUTO_TOTAL"], ascending=[True, False])
```

```python
# Filtrar linhas com rank < 5 e exibir apenas as colunas desejadas
filtro = dados_2[dados_2["rank"] <= 5][["SECAO_PRODUTO", "CATEGORIA_PRODUTO", "FATURAMENTO_BRUTO_TOTAL", "rank"]]
filtro
```
